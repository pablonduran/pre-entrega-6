# Demostracion del Orquestador Multi-Agente Jerarquico

Este notebook ejecuta paso a paso el StateGraph con Supervisor, Investigador, Analista y Validador.

```python
from langchain_core.messages import HumanMessage
from graph import construir_grafo_orquestador

orquestador = construir_grafo_orquestador()
print("Grafo compilado correctamente")
```

```python
consulta = (
    "Necesito analizar la infraestructura de servidores cloud de ana.gomez@hotmail.com, "
    "calcular su costo mensual total y determinar cuanto ahorraria anualmente si le aplicamos "
    "un 15% de descuento por contrato anual."
)

estado_inicial = {
    "messages": [HumanMessage(content=consulta)],
    "siguiente_paso": "investigador",
    "datos_investigacion": {},
    "analisis_computo": {},
    "validacion": {},
    "tarea_completada": False,
    "contador_pasos": 0,
    "historial_delegaciones": [],
}

print("Ejecutando orquestador...")
resultado = orquestador.invoke(estado_inicial, {"recursion_limit": 15})

print("\nRespuesta final sintetizada:")
print(resultado["messages"][-1].content)
```
